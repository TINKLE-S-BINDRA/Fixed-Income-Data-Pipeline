```python
def nelson_siegel(t, beta0, beta1, beta2, tau):
    return (beta0 + beta1 * ((1 - np.exp(-t/tau)) / (t/tau)) + beta2 * (((1 - np.exp(-t/tau)) / (t/tau)) - np.exp(-t/tau)))
```

```python
t = df["TTM_years"].values
y = df["Ask Yield"].values
```

```python
import numpy as np
from scipy.optimize import curve_fit
initial_guess = [np.mean(y), -1, 1, 1]
params, _ = curve_fit(nelson_siegel, t, y, p0=initial_guess, maxfev=10000)
t_smooth = np.linspace(0.01, 5, 100)
y_smooth = nelson_siegel(t_smooth, *params)
```

```python
plt.plot(t_smooth, y_smooth, label="Yield Curve", linewidth=2)
plt.title("Yield Curve")
plt.xlabel("Time to Maturity (Years)")
plt.ylabel("Yield (%)")
plt.grid(True)
plt.legend()
plt.show()
```
